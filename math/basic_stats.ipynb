```python
# lets import numpy
import numpy as np
import scipy.stats as stats
heights = np.random.randint(150, 200, size=100)

# assume that heights are normal distribution
mean = np.mean(heights)
std_dev = np.std(heights)

# find the probability of height more then 150
probability = 1 - stats.norm.cdf(10, loc=mean, scale=std_dev)
# number of people between 150 and 160
people_between = stats.norm.cdf(150, loc=mean, scale=std_dev) - stats.norm.cdf(190, loc=mean, scale=std_dev)

people_between
```

```python
# create normal distribution with different mean and variance
mean = 150
std_dev = 10
heights = np.random.normal(loc=mean, scale=std_dev, size=100)

# plot it
import matplotlib.pyplot as plt
plt.hist(heights, bins=10, alpha=0.7)
plt.title("Height Distribution")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
# lets create normal distribution random data 
import numpy as np
np.random.seed(0)
data = np.random.normal(loc=0, scale=1, size=1000)

# lets calculate basic statistics
mean = np.mean(data)
median = np.median(data)
std_dev = np.std(data)

mean, median, std_dev
```

```python
# lets plot the data
import matplotlib.pyplot as plt
plt.hist(data, bins=30, alpha=0.5, color='b')
plt.title('Histogram of Normally Distributed Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
# lets create skewed data
skewed_data = np.random.exponential(scale=1, size=1000)
# lets plot histogram
plt.hist(skewed_data, bins=30, alpha=0.5, color='r')
plt.title('Histogram of Skewed Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()  
```

```python
# lets calculate kurtosis
import scipy.stats as stats
kurtosis = stats.kurtosis(skewed_data)
kurtosis
```

```python
# lets create a random variable name as income
income = np.random.rand(1000) * 100


# lets create 30 sample of size 30
income_samples = [np.random.choice(income, size=30, replace=False) for _ in range(30)]
# lets calculate the mean of each sample
income_means = [np.mean(sample) for sample in income_samples]
print('Mean of each sample is :', income_means)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# plot KDE

plt.figure(figsize=(10, 6))
sns.kdeplot(income_means, fill=True)
plt.title('KDE of Income Means')
plt.xlabel('Income Mean')
plt.ylabel('Density')
plt.show()
```

```python
# to apply the Central Limit Theorem
mean_income = np.mean(income_means)
std_income = np.std(income_means)

np.random.normal(mean_income, std_income/np.sqrt(30), size=1000)
```

```python
# lets create a random normal distribution
simulated_income = np.random.normal(mean_income, std_income/np.sqrt(30), size=1000)


# lets apply standard normal transformation
standardized_income = (simulated_income - mean_income) / (std_income/np.sqrt(30))
standardized_income
```
