```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller,acf,pacf
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
# sample time series data with trend and seasonality

np.random.seed(42)
time = np.arange(100)
trend = time * 0.5
seasonality = 10 * np.sin(2 * np.pi * time / 12)
noise = np.random.normal(0, 2, size=100)
data = trend + seasonality + noise + 50
ts_data = pd.Series(data)

df = pd.DataFrame({'value': ts_data})
df.index = pd.date_range(start='2025-01-01', periods=len(df), freq='ME')
```

```python
# plot using seaborn along with trend and seasonality
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x=df.index, y='value')
plt.title('Sample Time Series Data with Trend and Seasonality')
plt.xlabel('Date')
plt.ylabel('Value')
plt.show()
```
