```python
from nltk.tokenize import word_tokenize
tokenized_docs = []
for doc in corpus:
    tokens = word_tokenize(doc)
    tokenized_docs.append(tokens)
for i, doc in enumerate(tokenized_docs):
    print(f"Document {i+1} tokens: {doc[:10]}...")  # Print first 10 tokens of each document
```

```python
# lowercase
lower_cased_docs = []
for doc in tokenized_docs:
    lower_cased_docs.append([token.lower() for token in doc])

for i, doc in enumerate(lower_cased_docs):
    print(f"Document {i+1} lowercased tokens: {doc[:10]}...")  # Print first 10 tokens of each document
```

```python
no_punctuation_docs = []

for tokens in lower_cased_docs:
    filtered_tokens = [token for token in tokens if token.isalnum()]
    no_punctuation_docs.append(filtered_tokens)

for i, doc in enumerate(no_punctuation_docs):
    print(f"Document {i+1} no punctuation tokens: {doc[:10]}...")  # Print first 10 tokens of each document
```

```python
# stop words removal
from nltk.corpus import stopwords
stop_words = set(stopwords.words("english"))
filtered_docs = []
for doc in no_punctuation_docs:
    filtered_tokens = [token for token in doc if token not in stop_words]
    filtered_docs.append(filtered_tokens)

for i, doc in enumerate(filtered_docs):
    print(f"Document {i+1} filtered tokens: {doc[:10]}...")  # Print first 10 tokens of each document
```

```python
# add also in stop words removal
stop_words.add("also")
no_stop_including_also_docs = []
for doc in no_punctuation_docs:
    filtered_tokens = [token for token in doc if token not in stop_words]
    no_stop_including_also_docs.append(filtered_tokens)

for i, doc in enumerate(no_stop_including_also_docs):
    print(f"Document {i+1} no stop including 'also' tokens: {doc[:10]}...")  # Print first 10 tokens of each document
```

# Difference between stemming and lemmatization
Stemming and lemmatization are both techniques used in natural language processing (NLP) to reduce words to their base or root form. However, they differ in their approach and the results they produce.

1. Stemming:
   - Stemming is a rule-based process that removes suffixes from words to obtain their root form.
   - It often involves simple heuristics and may not produce real words.
   - For example, the stemmer might reduce "running," "runner," and "ran" to "run."
   - Stemming is generally faster but less accurate than lemmatization.

2. Lemmatization:
   - Lemmatization is a more advanced process that considers the context and meaning of words.
   - It uses a vocabulary and morphological analysis to return the base form (lemma) of a word.
   - For example, lemmatization would reduce "better" to "good" and "running" to "run."
   - Lemmatization is more accurate but also more computationally intensive than stemming.

In summary, stemming is a faster but less accurate method of reducing words to their root form, while lemmatization is more accurate and considers the context and meaning of words.

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

stemmed_word = []
for doc in no_stop_including_also_docs:
    stemmed_tokens = [stemmer.stem(token) for token in doc]
    stemmed_word.append(stemmed_tokens)

for i, doc in enumerate(stemmed_word):
    print(f"Document {i+1} stemmed tokens: {doc[:10]}...")  # Print first 10 tokens of each document
```

### Advantages-
- Faster and simpler: It's computationally less expensive than other techniques like lemmatization.
- Reduces vocabulary size: By grouping words with similar stems, it helps manage large datasets more efficiently.

### Disadvantages
- The most prevelant one is Oversimplification: Can lead to incorrect stems, like transforming "cars" and "care" into the same stem "car".
- Meaning loss: Removing suffixes can alter the word's meaning, causing confusion. For eg- in our case the word intimate has been replaced with the word intim.
- Not context-aware: Doesn't consider different grammatical roles a word can play within a sentence.


```python
# Lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_word = []
for doc in no_stop_including_also_docs:
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in doc]
    lemmatized_word.append(lemmatized_tokens)

for i, doc in enumerate(lemmatized_word):
    print(f"Document {i+1} lemmatized tokens: {doc[:10]}...")  # Print first 10 tokens of each document
```
