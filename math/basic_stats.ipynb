```python
# lets create a random variable name as income
income = np.random.rand(1000) * 100


# lets create 30 sample of size 30
income_samples = [np.random.choice(income, size=30, replace=False) for _ in range(30)]
# lets calculate the mean of each sample
income_means = [np.mean(sample) for sample in income_samples]
print('Mean of each sample is :', income_means)
```

```python
# to apply the Central Limit Theorem
mean_income = np.mean(income_means)
std_income = np.std(income_means)

np.random.normal(mean_income, std_income/np.sqrt(30), size=1000)
```

```python
# lets create a random normal distribution
simulated_income = np.random.normal(mean_income, std_income/np.sqrt(30), size=1000)


# lets apply standard normal transformation
standardized_income = (simulated_income - mean_income) / (std_income/np.sqrt(30))
standardized_income
```

```python
## lets implement q-q plot from scratch
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats 
# generate random normal data
income = np.random.normal(loc=0, scale=1, size=1000)
# generate theoretical quantiles
theoretical_quantiles = np.sort(np.random.normal(loc=0, scale=1, size=1000))
# sort the data and compute percentiles
income = np.sort(income)
percentiles = np.linspace(0, 100, len(income))
# compute empirical quantiles
empirical_quantiles = np.percentile(income, percentiles)

# plot the q-q plot
plt.figure(figsize=(8, 8))
plt.scatter(theoretical_quantiles, empirical_quantiles, color='blue', s=10)
plt.plot([-3, 3], [-3, 3], color='red', linestyle='--')
plt.title('Q-Q Plot')
plt.xlabel('Theoretical Quantiles')
plt.ylabel('Empirical Quantiles')
plt.xlim([-3, 3])
plt.ylim([-3, 3])
plt.grid()
plt.show()
```

```python
# lets create income variable between 10k to 20k which may be normal distribution
income = np.random.normal(loc=15000, scale=2000, size=1000)
income = np.clip(income, 10000, 20000)

```

```python
# lets apply ks test to check if income is normal distribution
ks_statistic, p_value = stats.kstest(income, 'norm', args=(np.mean(income), np.std(income)))
ks_statistic, p_value
```

```python
# lets create a variable called sales which is not normal distribution
sales = np.random.exponential(scale=3000, size=1000)
sales = np.clip(sales, 1000, 20000)
# lets apply ks test to check if sales is normal distribution
ks_statistic, p_value = stats.kstest(sales, 'norm', args=(np.mean(sales), np.std(sales)))
print('KS Statistic:', ks_statistic)
print('P-value:', p_value)  

# lets plot it
plt.hist(sales, bins=30, alpha=0.5, color='g')
plt.title('Histogram of Sales Data')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
# lets create fake data from banking domain
import numpy as np
np.random.seed(42)
n_samples = 1000
ages = np.random.randint(18, 70, size=n_samples)
incomes = np.random.normal(loc=50000, scale=15000, size=n_samples)
incomes = np.clip(incomes, 20000, 150000)
balances = np.random.normal(loc=20000, scale=10000, size=n_samples)
balances = np.clip(balances, 0, 100000)
# lets create a categorical variable for account type
account_types = np.random.choice(['savings', 'checking', 'business'], size=n_samples, p=[0.5, 0.4, 0.1])        
# lets create a binary variable for churn
churn = np.random.choice([0, 1], size=n_samples, p=[0.8, 0.2])        
# lets create a structured array
data = np.core.records.fromarrays([ages, incomes, balances, account_types, churn],
                                  names='age, income, balance, account_type, churn')    

import pandas as pd
df = pd.DataFrame(data)
df.head()
```

```python
# let see the distribution of income in case of churn and non churn customers
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.kdeplot(df[df['churn'] == 0]['income'], label='Non-Churn')
sns.kdeplot(df[df['churn'] == 1]['income'], label='Churn')
plt.title('Income Distribution by Churn Status')
plt.xlabel('Income')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
# lets check it income is normal distribution or not using ks test
ks_statistic, p_value = stats.kstest(df['income'], 'norm', args=(np.mean(df['income']), np.std(df['income'])))
print('KS Statistic:', ks_statistic)
print('P-value:', p_value)

if p_value < 0.05:
    print("Reject the null hypothesis: Data is not normally distributed")
else:
    print("Fail to reject the null hypothesis: Data is normally distributed")
```

```python
# sicne income is not normal distribution we can apply log transformation
df['log_income'] = np.log(df['income'])
plt.figure(figsize=(10, 6))
sns.kdeplot(df[df['churn'] == 0]['log_income'], label='Non-Churn')
sns.kdeplot(df[df['churn'] == 1]['log_income'], label='Churn')
plt.title('Log Income Distribution by Churn Status')
plt.xlabel('Log Income')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
# lets compute few stats based on income ( since it is normal distribution)
print("Income Statistics:")
print("Mean:", df['income'].mean())
print("Median:", df['income'].median())
print("Standard Deviation:", df['income'].std())

# what is probability of income more than 70k
mean_income = df['income'].mean()
std_income = df['income'].std()
probability_above_70k = 1 - stats.norm.cdf(2000, loc=mean_income, scale=std_income)
probability_above_70k
```

```python
import sklearn.datasets as datasets
# load iris dataset
iris = datasets.load_iris()
X = iris.data
y = iris.target
# convert to dataframe
df_iris = pd.DataFrame(data=X, columns=iris.feature_names)
df_iris['target'] = y
df_iris.head()
```

```python
# lets plot pair plot
sns.pairplot(df_iris, hue='target', vars=iris.feature_names)
plt.show()
```

```python
# lets create python function for iris dataset rule based rule

def predict_iris(row):
    if row['petal length (cm)'] < 2:
        return 0  # Setosa
    elif row['petal length (cm)'] < 5:
        return 1  # Versicolor
    else:
        return 2  # Virginica
    
# --- IGNORE ---
# pass one row to test your predict_iris function
predict_iris(df_iris.iloc[0])
# --- IGNORE ---    
df_iris['predicted'] = df_iris.apply(predict_iris, axis=1)
# lets check accuracy
accuracy = (df_iris['predicted'] == df_iris['target']).mean()
accuracy
```
