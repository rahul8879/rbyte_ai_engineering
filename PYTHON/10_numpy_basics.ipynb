```python
def parse_row(row):
    try:
        return (int(row[0]), row[1], float(row[2]), int(row[3]))
    except ValueError:

        print('isnide alue:',row)
        return None  # drop or log the bad row
```

```python
cleaned = [parse_row(row) for row in data if parse_row(row)]
cleaned
```

```python
arr = np.array(cleaned, dtype=[('id', 'i4'), ('name', 'U10'), ('price', 'f4'), ('qty', 'i4')])
arr
```

```python
# Total revenue per product
revenue = arr['price'] * arr['qty']
print("Revenue:", revenue)

# Best-selling item
best_idx = np.argmax(revenue)
print("Top product:", arr[best_idx]['name'])
```

```python
# # Case Study: Inventory Optimization for a Retail Store
# 🧩 Problem:
# A retail chain has inventory data for 5 stores and 4 products. You want to:

# Calculate the total inventory across stores.

# Apply a restocking threshold (broadcasting).

# Calculate cost of inventory per product (vectorized multiplication).
```

```python
inventory = np.array([
    [20, 35, 10, 50],
    [30, 40, 20, 60],
    [15, 30, 25, 40],
    [50, 20, 30, 70],
    [10, 45, 15, 55]
])

# Unit prices for 4 products
prices = np.array([100, 150, 200, 250])
```

```python
#  Step 2: Total Inventory for Each Product
inventory.sum(axis=1)
```

```python
# 🚨 Step 3: Find Low Stock (Broadcasting threshold)
```

```python
threshold = 20
```

```python
low_stock_mask = inventory < threshold
low_stock_mask
print("Low Stock (less than 20):\n", low_stock_mask)
```

```python
# 💰 Step 4: Compute Total Inventory Cost per Store
inventory_value = inventory * prices 
store_inventory_cost = inventory_value.sum(axis=1)
print("Total Inventory Cost per Store:", store_inventory_cost)
```

```python
discount_mask = inventory > 40
discounted_inventory_value = np.where(discount_mask, inventory * prices * 0.9, inventory * prices)
print("Discounted Inventory Value:\n", discounted_inventory_value)
```
