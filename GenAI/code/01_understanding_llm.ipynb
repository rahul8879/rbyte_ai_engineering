```python
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
import json

client = OpenAI()
```

```python
# define tools
tools = [
    {
        "type": "function",
        "function": {
            "name": "get_weather",
            "description": "Get current weather for a given city",
            "parameters": {
                "type": "object",
                "properties": {
                    "city": {
                        "type": "string",
                        "description": "The city name, e.g. Mumbai"
                    },
                    "unit": {
                        "type": "string",
                        "enum": ["celsius", "fahrenheit"],
                        "description": "Temperature unit"
                    }
                },
                "required": ["city"]
            }
        }
    }
]
```

```python
messages = [
    {
        "role": "user",
        "content": "What's the weather like in Mumbai?"
    }
]

response = client.chat.completions.create(
    model="gpt-4o",
    messages=messages,
    tools=tools,
    tool_choice="auto"  # Let the model decide when to call
)

response_message = response.choices[0].message
print(response_message)
```
