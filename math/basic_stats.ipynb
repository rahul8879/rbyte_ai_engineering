```python
# lets import numpy
import numpy as np
import scipy.stats as stats
heights = np.random.randint(150, 200, size=100)

# assume that heights are normal distribution
mean = np.mean(heights)
std_dev = np.std(heights)

# find the probability of height more then 150
probability = 1 - stats.norm.cdf(10, loc=mean, scale=std_dev)
# number of people between 150 and 160
people_between = stats.norm.cdf(150, loc=mean, scale=std_dev) - stats.norm.cdf(190, loc=mean, scale=std_dev)

people_between
```

```python
# create normal distribution with different mean and variance
mean = 150
std_dev = 10
heights = np.random.normal(loc=mean, scale=std_dev, size=100)

# plot it
import matplotlib.pyplot as plt
plt.hist(heights, bins=10, alpha=0.7)
plt.title("Height Distribution")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.show()
```
