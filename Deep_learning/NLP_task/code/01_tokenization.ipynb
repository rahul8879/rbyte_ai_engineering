```python
import numpy as np

# Step 1: Create a simple corpus
corpus = [
    "I love cats",
    "Dogs are great pets", 
    "Cats and dogs are animals",
    "I have a pet cat",
    "My dog loves treats"
]
```

```python
import numpy as np

print("=== UNDERSTANDING TOKENIZATION, TOKEN IDs, AND EMBEDDINGS ===\n")



print("STEP 1: Raw Text")
print("-" * 20)
for i, text in enumerate(corpus):
    print(f"{i}: {text}")

# Step 2: Build vocabulary (THIS IS MODEL-SPECIFIC!)
print("\nSTEP 2: Build Vocabulary (Token-to-ID Mapping)")
print("-" * 50)
print("⚠️  IMPORTANT: This mapping is SPECIFIC to this model!")
print("   Different models will assign different IDs to same words\n")

def build_vocab(corpus):
    # Tokenize (simple word splitting)
    all_words = []
    for text in corpus:
        words = text.lower().split()
        all_words.extend(words)
    
    # Create vocabulary with special tokens
    vocab = {"<PAD>": 0, "<UNK>": 1}  # Special tokens
    
    # Add unique words from corpus
    unique_words = list(set(all_words))
    for i, word in enumerate(sorted(unique_words), start=2):  # Sort for consistency
        vocab[word] = i
    
    return vocab

vocab = build_vocab(corpus)

print("OUR MODEL'S VOCABULARY:")
for word, token_id in sorted(vocab.items(), key=lambda x: x[1]):
    print(f"  '{word}' → Token ID: {token_id}")

# Step 3: Show that different models have different mappings
print(f"\nSTEP 3: Different Models = Different Token IDs")
print("-" * 45)
print("Same word 'cat' in different models:")
print(f"  Our Model:    'cat' → {vocab['cat']}")
print(f"  GPT Model:    'cat' → 2245   (hypothetical)")
print(f"  BERT Model:   'cat' → 4829   (hypothetical)")
print(f"  Word2Vec:     'cat' → 156    (hypothetical)")
print("\n✅ Key Point: Token IDs are NOT universal - they're model-specific!")

# Step 4: Convert text to token IDs
def text_to_ids(text, vocab):
    words = text.lower().split()
    ids = []
    for word in words:
        ids.append(vocab.get(word, vocab["<UNK>"]))  # Use <UNK> for unknown words
    return ids

print(f"\nSTEP 4: Text → Token IDs")
print("-" * 25)
example_text = "I love cats"
token_ids = text_to_ids(example_text, vocab)
words = example_text.lower().split()

print(f"Text: '{example_text}'")
print(f"Tokenization: {words}")
print(f"Token IDs: {token_ids}")
print(f"Mapping:")
for word, tid in zip(words, token_ids):
    print(f"  '{word}' → {tid}")

# Step 5: Create embedding lookup table
print(f"\nSTEP 5: Embedding Lookup Table")
print("-" * 32)
print("Token IDs are just INDICES into an embedding table!")

vocab_size = len(vocab)
embedding_dim = 4  # Small for demo (real models use 256, 512, 768+)

# Initialize random embeddings (in real models, these are learned during training)
np.random.seed(42)  # For reproducible results
embedding_table = np.random.randn(vocab_size, embedding_dim)

print(f"Vocabulary size: {vocab_size}")
print(f"Embedding dimension: {embedding_dim}")
print(f"Embedding table shape: {embedding_table.shape}")
print("\nEmbedding Table (Token ID → Dense Vector):")
print("Token ID | Word      | Embedding Vector")
print("-" * 45)

id_to_word = {v: k for k, v in vocab.items()}
for token_id in range(min(8, vocab_size)):  # Show first 8 for brevity
    word = id_to_word[token_id]
    embedding = embedding_table[token_id]
    print(f"{token_id:8d} | {word:9s} | {embedding}")

# Step 6: Embedding lookup process
print(f"\nSTEP 6: Embedding Lookup Process")
print("-" * 33)

def get_embeddings(token_ids, embedding_table):
    """Convert token IDs to embeddings by looking up in the table"""
    return embedding_table[token_ids]

sentence = "I love cats"
token_ids = text_to_ids(sentence, vocab)
sentence_embeddings = get_embeddings(token_ids, embedding_table)

print(f"Sentence: '{sentence}'")
print(f"Token IDs: {token_ids}")
print(f"Embeddings shape: {sentence_embeddings.shape}")
print(f"This means: {len(token_ids)} tokens, each with {embedding_dim}-dimensional embedding")

print(f"\nDetailed breakdown:")
words = sentence.lower().split()
for i, (word, token_id) in enumerate(zip(words, token_ids)):
    embedding = sentence_embeddings[i]
    print(f"  '{word}' (ID: {token_id}) → {embedding}")

# Step 7: Show the complete pipeline
print(f"\nSTEP 7: Complete Pipeline Summary")
print("-" * 34)
print("Text → Tokenization → Token IDs → Embedding Lookup → Dense Vectors")
print("")
example = "My dog loves treats"
print(f"1. Text: '{example}'")

tokens = example.lower().split()
print(f"2. Tokens: {tokens}")

ids = text_to_ids(example, vocab)
print(f"3. Token IDs: {ids}")

embeddings = get_embeddings(ids, embedding_table)
print(f"4. Embeddings: shape {embeddings.shape}")

print(f"5. Dense vectors (what the model actually processes):")
for i, (token, tid, emb) in enumerate(zip(tokens, ids, embeddings)):
    print(f"   Position {i}: '{token}' → {emb}")

# Step 8: Key insights
print(f"\nSTEP 8: Key Insights")
print("-" * 18)
print("✅ Token IDs are just lookup indices - they have no inherent meaning")
print("✅ Token IDs are model-specific (not worldwide standards)")
print("✅ Embeddings are the dense vectors that capture semantic meaning")
print("✅ The model works with embeddings, not token IDs directly")
print("✅ Embedding table is learned during training (we used random values)")

print(f"\nWord2Vec Example:")
print("-" * 16)
print("If you used Word2Vec:")
print("1. Word2Vec has its own vocabulary and token-to-ID mapping")
print("2. 'cat' might get a completely different token ID")
print("3. But the embedding lookup process is the same")
print("4. Word2Vec embeddings are pre-trained to capture word relationships")

# Step 9: Demonstrate why we need both
print(f"\nSTEP 9: Why We Need Both Token IDs AND Embeddings")
print("-" * 48)
print("Token IDs alone are useless for understanding:")
cat_id = vocab['cat']
dog_id = vocab['dogs']
print(f"  'cat' → {cat_id}")
print(f"  'dogs' → {dog_id}")
print(f"  Difference: {abs(cat_id - dog_id)} (meaningless number)")

print(f"\nEmbeddings capture semantic relationships:")
cat_emb = embedding_table[cat_id]
dog_emb = embedding_table[dog_id]
# Calculate cosine similarity
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

similarity = cosine_similarity(cat_emb, dog_emb)
print(f"  'cat' embedding: {cat_emb}")
print(f"  'dogs' embedding: {dog_emb}")
print(f"  Cosine similarity: {similarity:.3f}")
print(f"  (In a trained model, 'cat' and 'dog' would have high similarity)")

print(f"\n" + "="*60)
print("FINAL ANSWER TO YOUR QUESTION:")
print("="*60)
print("❌ Token IDs are NOT fixed worldwide")
print("✅ Each model has its own vocabulary and token-ID mapping") 
print("✅ Embedding lookup uses these model-specific token IDs as indices")
print("✅ The embedding table contains the learned semantic representations")
print("✅ You need BOTH: Token IDs (for indexing) + Embeddings (for meaning)")
```
