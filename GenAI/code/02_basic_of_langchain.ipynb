```python
from langchain.chains import LLMChain
from langchain_core.prompts import ChatPromptTemplate

prompt_text ="{query}"
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant.Please help me to learn given topic"), 
    ("human", prompt_text)])

llm_chain = LLMChain(llm=llm, prompt=prompt_template)
```

```python
llm_chain.invoke({"query": "what is 2+ 2"})
```

```python
#LCEL : use this only on production

llm_chain = (prompt_template | llm)

llm_chain.invoke({"query": "Who is PM of India?"})
```

```python
from langchain_core.prompts import PromptTemplate,ChatPromptTemplate

pt = PromptTemplate(
    input_variable=['name','topic'],
    template="""Your name is {name}. You are a helpful assistant. Please help me to learn about {topic}."""
)

result = pt.format_prompt(name="Rbyte.ai", topic="AI/ML")

print(result)

llm_chain = (pt | llm)
llm_chain.invoke({"name": "Rbyte.ai", "topic": "AI/ML"})
```
